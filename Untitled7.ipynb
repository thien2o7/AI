```python
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl
import matplotlib.pyplot as plt

# ========== 1. Định nghĩa các biến ngôn ngữ ==========

# --- Đầu vào ---
# Quãng đường (km)
distance = ctrl.Antecedent(np.arange(0, 51, 1), 'distance')
# Tình trạng giao thông (%)
traffic = ctrl.Antecedent(np.arange(0, 101, 1), 'traffic')
# Mức cầu (%)
demand = ctrl.Antecedent(np.arange(0, 101, 1), 'demand')
# Điều kiện thời tiết (thang điểm 0-10)
weather = ctrl.Antecedent(np.arange(0, 11, 1), 'weather')
# Đánh giá của khách hàng (1-5)
rating = ctrl.Antecedent(np.arange(1, 5.1, 0.1), 'rating')
# Đúng giờ (%)
punctuality = ctrl.Antecedent(np.arange(0, 101, 1), 'punctuality')

# --- Đầu ra ---
# Giá chuyến đi (đơn vị quy ước 0-100)
price = ctrl.Consequent(np.arange(0, 101, 1), 'price')
# Điểm thưởng cho tài xế (0-10)
bonus = ctrl.Consequent(np.arange(0, 11, 1), 'bonus')

# ========== 2. Định nghĩa hàm liên thuộc ==========

# Quãng đường
distance['Short'] = fuzz.trimf(distance.universe, [0, 0, 3])
distance['Medium'] = fuzz.trimf(distance.universe, [2, 5, 8])
distance['Long'] = fuzz.trimf(distance.universe, [6, 13, 20])
distance['Very_Long'] = fuzz.trimf(distance.universe, [15, 32.5, 50])

# Giao thông
traffic['Low'] = fuzz.trimf(traffic.universe, [0, 0, 30])
traffic['Medium'] = fuzz.trimf(traffic.universe, [20, 50, 70])
traffic['High'] = fuzz.trimf(traffic.universe, [60, 100, 100])

# Mức cầu
demand['Low'] = fuzz.trimf(demand.universe, [0, 0, 30])
demand['Medium'] = fuzz.trimf(demand.universe, [20, 50, 70])
demand['High'] = fuzz.trimf(demand.universe, [60, 100, 100])

# Thời tiết (0: tốt, 5: trung bình, 10: xấu)
weather['Good'] = fuzz.trimf(weather.universe, [0, 0, 4])
weather['Moderate'] = fuzz.trimf(weather.universe, [3, 5, 7])
weather['Bad'] = fuzz.trimf(weather.universe, [6, 10, 10])

# Đánh giá khách hàng
rating['Poor'] = fuzz.trimf(rating.universe, [1, 1, 2.5])
rating['Average'] = fuzz.trimf(rating.universe, [2, 3.5, 4])
rating['Good'] = fuzz.trimf(rating.universe, [3.5, 5, 5])

# Đúng giờ
punctuality['Late'] = fuzz.trimf(punctuality.universe, [0, 0, 50])
punctuality['On_Time'] = fuzz.trimf(punctuality.universe, [40, 60, 80])
punctuality['Early'] = fuzz.trimf(punctuality.universe, [70, 100, 100])

# Giá chuyến đi
price['Low'] = fuzz.trimf(price.universe, [0, 0, 30])
price['Medium'] = fuzz.trimf(price.universe, [20, 50, 70])
price['High'] = fuzz.trimf(price.universe, [60, 80, 100])
price['Very_High'] = fuzz.trimf(price.universe, [80, 100, 100])

# Điểm thưởng
bonus['None'] = fuzz.trimf(bonus.universe, [0, 0, 2])
bonus['Few'] = fuzz.trimf(bonus.universe, [1, 3, 5])
bonus['Moderate'] = fuzz.trimf(bonus.universe, [4, 6, 8])
bonus['High'] = fuzz.trimf(bonus.universe, [7, 10, 10])

# ========== 3. Định nghĩa các luật mờ (dựa trên tài liệu) ==========

# --- Luật cho giá chuyến đi (1-10) ---
rule1 = ctrl.Rule(distance['Short'] & traffic['Low'] & demand['Low'], price['Low'])
rule2 = ctrl.Rule(distance['Short'] & traffic['Medium'] & demand['High'], price['Medium'])
rule3 = ctrl.Rule(distance['Medium'] & traffic['High'] & demand['High'], price['High'])
rule4 = ctrl.Rule(distance['Long'] & traffic['Medium'] & weather['Good'], price['Medium'])
rule5 = ctrl.Rule(distance['Long'] & traffic['High'] & weather['Bad'], price['Very_High'])
rule6 = ctrl.Rule(distance['Very_Long'] & traffic['High'] & demand['High'], price['Very_High'])
rule7 = ctrl.Rule(distance['Medium'] & traffic['Low'] & demand['Low'], price['Medium'])
rule8 = ctrl.Rule(distance['Short'] & traffic['High'] & weather['Bad'], price['High'])
rule9 = ctrl.Rule(distance['Very_Long'] & weather['Bad'], price['Very_High'])
rule10 = ctrl.Rule(distance['Medium'] & traffic['Medium'] & weather['Bad'], price['Medium'])

# --- Luật cho điểm thưởng (11-20) ---
rule11 = ctrl.Rule(rating['Good'] & punctuality['Early'], bonus['High'])
rule12 = ctrl.Rule(rating['Average'] & punctuality['On_Time'], bonus['Moderate'])
rule13 = ctrl.Rule(rating['Poor'] & punctuality['Late'], bonus['None'])
rule14 = ctrl.Rule(distance['Long'] & traffic['High'] & punctuality['On_Time'], bonus['High'])
rule15 = ctrl.Rule(distance['Medium'] & traffic['Medium'] & rating['Good'], bonus['Moderate'])
rule16 = ctrl.Rule(rating['Poor'] & punctuality['Late'], bonus['None'])   # trùng ý nghĩa rule13
rule17 = ctrl.Rule(distance['Very_Long'] & weather['Bad'] & rating['Good'], bonus['High'])
rule18 = ctrl.Rule(distance['Short'] & rating['Average'] & punctuality['On_Time'], bonus['Few'])
rule19 = ctrl.Rule(distance['Long'] & traffic['High'] & punctuality['Late'], bonus['Few'])
rule20 = ctrl.Rule(distance['Medium'] & weather['Moderate'] & rating['Good'], bonus['Moderate'])

# ========== 4. Xây dựng hệ thống điều khiển ==========
price_ctrl = ctrl.ControlSystem([rule1, rule2, rule3, rule4, rule5, rule6, rule7, rule8, rule9, rule10])
bonus_ctrl = ctrl.ControlSystem([rule11, rule12, rule13, rule14, rule15, rule16, rule17, rule18, rule19, rule20])

price_sim = ctrl.ControlSystemSimulation(price_ctrl)
bonus_sim = ctrl.ControlSystemSimulation(bonus_ctrl)

# ========== 5. Mô phỏng với một bộ input cụ thể ==========
# Ví dụ: quãng đường 7 km, giao thông 65%, nhu cầu 70%, thời tiết 8 (xấu), đánh giá 4.5 (tốt), đúng giờ 90% (sớm)

price_sim.input['distance'] = 18
price_sim.input['traffic'] = 85
price_sim.input['demand'] = 90
price_sim.input['weather'] = 9

bonus_sim.input['distance'] = 18
bonus_sim.input['traffic'] = 85
bonus_sim.input['weather'] = 9
bonus_sim.input['rating'] = 4.8
bonus_sim.input['punctuality'] = 85

# Tính toán
price_sim.compute()
bonus_sim.compute()

# In kết quả
print("===== KẾT QUẢ SUY LUẬN MỜ =====")
print(f"Giá chuyến đi (quy đổi): {price_sim.output['price']:.2f}")
print(f"Điểm thưởng tài xế: {bonus_sim.output['bonus']:.2f}")

# ========== 6. Trực quan hóa (tuỳ chọn) ==========
# Hiển thị hàm liên thuộc của một vài biến
distance.view()
traffic.view()
weather.view()
price.view()
bonus.view()
plt.show()
```
